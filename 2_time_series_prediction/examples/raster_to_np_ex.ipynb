```python
import rasterio
import numpy as np

with rasterio.open("./data/raster_ex.tif") as src:
    array = src.read(1)  # first band
    print(array.shape)
    np.save("./data/raster_ex.npy", array)
```

```python
import matplotlib.pyplot as plt

plt.imshow(array, cmap="gray", origin="upper")
plt.title("Rasterized Water Mask")
plt.colorbar(label='Label (0: background, 1: water)')
plt.show()
```
